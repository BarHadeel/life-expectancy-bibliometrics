```python
!pip install xlrd -q

import pandas as pd, math
from google.colab import files

# 1. Upload your three savedrecs .xls files when prompted
uploaded = files.upload()

# 2. Merge and verify
dfs = [pd.read_excel(f) for f in uploaded.keys()]
df = pd.concat(dfs, ignore_index=True)
print("Total records:", len(df))
print("Unique WoS IDs:", df['UT (Unique WOS ID)'].nunique())  # should equal total

# 3. Draw the random sample (random_state=42 reproduces my exact sample)
sample = df.sample(100, random_state=42).reset_index(drop=True)
sample['relevance'] = ''   # fill with R / N / A manually
sample[['Article Title','Source Title','Author Keywords','Abstract','relevance']] \
    .to_csv('sample100_to_screen.csv', index=False)
files.download('sample100_to_screen.csv')

# 4. After screening, set r = number of Relevant records, then run:
r, n, z = 40, 100, 1.96          # replace 40 with your count
p = r/n
den = 1 + z*z/n
centre = (p + z*z/(2*n)) / den
hw = z*math.sqrt(p*(1-p)/n + z*z/(4*n*n)) / den
print(f"Relevant: {p:.0%}, Wilson 95% CI [{centre-hw:.1%}, {centre+hw:.1%}]")
print(f"Implied relevant among {len(df)}: {p*len(df):.0f} "
      f"[{(centre-hw)*len(df):.0f}, {(centre+hw)*len(df):.0f}]")

# 5. Citation check: top-cited added records
tc = df['Times Cited, WoS Core'].fillna(0).astype(int)
print(df.assign(TC=tc).nlargest(15,'TC')[['Article Title','Source Title','TC']].to_string())
print("Added records with ≥278 citations:", (tc>=278).sum())
```
